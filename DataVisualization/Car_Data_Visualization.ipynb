```python
# Import libraries

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
car_df = pd.read_csv('car_price_dataset.csv')
```

```python
car_df.info()
```

```python
car_df.nunique()
```

```python
sns.catplot(data = car_df, x = 'Brand', kind = 'count', aspect = 2)

plt.show()
```

```python
sns.catplot(data = car_df, x = 'Brand' , y = 'Price', kind = 'box', aspect = 2)

plt.show()
```

```python
sns.catplot(data = car_df, x = 'Brand' , y = 'Price', col = 'Fuel_Type', kind = 'violin', col_wrap = 2, aspect = 2)

plt.show()
```

```python
sns.lmplot(data = car_df, x = 'Mileage', y = 'Price', col = 'Brand', hue = 'Fuel_Type', col_wrap = 3)

plt.show()
```
